## 4. Data Cleaning

```python
df.shape
```

```python
df = df.drop(columns=["Unnamed: 64"])
```

```python
df.shape
```

```python
raw_df = pd.read_csv("players_3120.csv")

df["Overall rating"] = raw_df["Overall rating"].str.extract(r"(\d+)")[0].astype(float)
```

```python
df["Overall rating"].dtype
```

```python
df["Overall rating"].isnull().sum()
```

```python
df["Potential"] = df["Potential"].str.extract(r"(\d+)")[0].astype(float)
```

```python
df["Potential"].dtype
```

```python
df["Potential"].isnull().sum()
```

```python
df.dtypes
```

## 6. Data Visualization

```python
position_counts = df.groupby("Best position").size()

position_counts.plot(kind="bar")

plt.title("Number of Players by Position")
plt.xlabel("Position")
plt.ylabel("Number of Players")

plt.show()
```

**Finding:** CAM is the most common position with 614 players while CF is the least common with 14 players

```python
plt.hist(df["Overall rating"], bins=15)

plt.title("Distribution of Overall Ratings")
plt.xlabel("Overall Rating")
plt.ylabel("Number of Players")

plt.show()
```

**Finding:** The Overall ratings range from 47 to 91, with an average rating of 73.22.

```python
plt.scatter(df["Overall rating"], df["Potential"])

plt.title("Overall Rating vs Potential")
plt.xlabel("Overall Rating")
plt.ylabel("Potential")

plt.show()
```

**Finding:** Overall rating and Potential have a moderate positive relationship, with a correlation of approximately 0.546.

## 7. Key Findings

1. CAM is the most common position in the dataset with 614 players, while CF is the least common with 14 players.

2. The average Overall rating is 73.22, with ratings ranging from 47 to 91.

3. Overall rating and Potential have a moderate positive relationship, with a correlation of approximately 0.546.

4. CF players have the highest average Overall rating at 78.43, while RM players have the lowest average at 71.41. However, CF has only 14 players, so this average is based on a relatively small group.

## 8. Conclusion

This analysis explored a football player statistics dataset containing 3120 players.

The dataset was cleaned using Pandas by removing an empty column and converting the Overall rating and Potential columns into numerical values.

The data was then explored using descriptive statistics, filtering, sorting, grouping, and summarization. Three meaningful visualizations were created using Matplotlib.

The analysis showed differences in player positions the distribution of Overall ratings, and a moderate positive relationship between Overall rating and Potential.
